```python
comparison_mean = df_clean.groupby("Class").mean(numeric_only=True).T

comparison_mean
```

```python
mean_comparison = df_clean.groupby("Class").mean(numeric_only=True).T

mean_comparison["absolute_difference"] = (
    mean_comparison[1] - mean_comparison[0]
).abs()

mean_comparison.sort_values("absolute_difference", ascending=False)
```

## Statistical Analysis — Initial Findings

The comparison between fraudulent and non-fraudulent transactions shows important differences in the mean values of several features.

V14, V3, V17, V12 and V10 show some of the largest differences in mean values between the two classes.

The average transaction amount is also higher for fraudulent transactions than for non-fraudulent transactions in this dataset.

However, differences in mean values alone are not sufficient to determine feature importance. The distributions and variability of these features need to be investigated during the Exploratory Data Analysis (EDA).
